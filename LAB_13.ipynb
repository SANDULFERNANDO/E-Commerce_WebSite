```python
import os, random, shutil
from pathlib import Path

random.seed(42)

src_dir = "/content/dogs-:contentReference[oaicite:5]{index=5}.0.jpg, dog.0.jpg are

base_dir = "/content/data/cats_dogs"
train_dir = f"{base_dir}/train"
val_dir = f"{base_dir}/validation"

for split_dir in [train_dir, val_dir]:
    Path(f"{split_dir}/cat").mkdir(parents=True, exist_ok=True)
    Path(f"{split_dir}/dog").mkdir(parents=True, exist_ok=True)

cats = [f for f in os.listdir(src_dir) if f.startswith("cat.")]
dogs = [f for f in os.listdir(src_dir) if f.startswith("dog.")]

random.shuffle(cats)
random.shuffle(dogs)

# Use small subset like the lab idea (example: 1000 per class train, 500 per class val)
train_cats, val_cats = cats[:1000], cats[1000:1500]
train_dogs, val_dogs = dogs[:1000], dogs[1000:1500]

def copy_files(file_list, label, target_folder):
    for fname in file_list:
        shutil.copy(os.path.join(src_dir, fname), os.path.join(target_folder, label, fname))

copy_files(train_cats, "cat", train_dir)
copy_files(train_dogs, "dog", train_dir)
copy_files(val_cats, "cat", val_dir)
copy_files(val_dogs, "dog", val_dir)

print("Train cats:", len(os.listdir(f"{train_dir}/cat")))
print("Train dogs:", len(os.listdir(f"{train_dir}/dog")))
print("Val cats:", len(os.listdir(f"{val_dir}/cat")))
print("Val dogs:", len(os.listdir(f"{val_dir}/dog")))
```

```python
import numpy as np
import matplotlib.pyplot as plt
import os
from tensorflow import keras
from tensorflow.keras.models import Model
from tensorflow.keras.layers import Dense, GlobalAveragePooling2D, Dropout
from tensorflow.keras.preprocessing.image import ImageDataGenerator
from tensorflow.keras.applications import VGG16, ResNet50, MobileNetV2
from tensorflow.keras.applications.vgg16 import preprocess_input as vgg16_preprocess
from tensorflow.keras.applications.resnet50 import preprocess_input as resnet50_preprocess
from tensorflow.keras.applications.mobilenet_v2 import preprocess_input as mobilenet_preprocess
from tensorflow.keras.callbacks import EarlyStopping, ReduceLROnPlateau
```

```python
# Define paths (adjust to your dataset location)
train_dir = 'data/cats_dogs/train'
validation_dir = 'data/cats_dogs/validation'

# Image dimensions
IMG_SIZE = 224
BATCH_SIZE = 32

# Training data generator
train_datagen = ImageDataGenerator(
 preprocessing_function=vgg16_preprocess,
 rotation_range=40,
 width_shift_range=0.2,
 height_shift_range=0.2,
 shear_range=0.2,
 zoom_range=0.2,
 horizontal_flip=True,
 fill_mode='nearest'
)

# Validation data generator
validation_datagen = ImageDataGenerator(
 preprocessing_function=vgg16_preprocess
)

# Create generators
train_generator = train_datagen.flow_from_directory(
 train_dir,
 target_size=(IMG_SIZE, IMG_SIZE),
 batch_size=BATCH_SIZE,
 class_mode='categorical'
)

validation_generator = validation_datagen.flow_from_directory(
 validation_dir,
 target_size=(IMG_SIZE, IMG_SIZE),
 batch_size=BATCH_SIZE,
 class_mode='categorical'
)

print(f'Training samples: {train_generator.samples}')
print(f'Validation samples: {validation_generator.samples}')
print(f'Classes: {train_generator.class_indices}')
```
